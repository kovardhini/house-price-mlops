```python
import pandas as pd
import numpy as np

train = pd.read_csv("../data/processed/train_cleaned.csv")
```

```python
train['TotalSF'] = train['TotalBsmtSF'] + train['1stFlrSF'] + train['2ndFlrSF']
```

```python
train['TotalBathrooms'] = (train['FullBath'] + (0.5 * train['HalfBath']) +
                             train['BsmtFullBath'] + (0.5 * train['BsmtHalfBath']))
```

```python
train['HouseAge'] = train['YrSold'] - train['YearBuilt']
```

```python
train['RemodAge'] = train['YrSold'] - train['YearRemodAdd']
```

```python
train['GarageAge'] = np.where(train['GarageYrBlt'] == 0, 0, train['YrSold'] - train['GarageYrBlt'])
```

```python
train['TotalPorchArea'] = (train['OpenPorchSF'] + train['EnclosedPorch'] +
                             train['3SsnPorch'] + train['ScreenPorch'] + train['WoodDeckSF'])
```

```python
train['TotalQualityScore'] = train['OverallQual'] * train['OverallCond']
```

```python
train['QualityLivingArea'] = train['OverallQual'] * train['GrLivArea']
```

```python
train['HasPool'] = (train['PoolArea'] > 0).astype(int)
train['HasGarage'] = (train['GarageArea'] > 0).astype(int)
train['HasFireplace'] = (train['Fireplaces'] > 0).astype(int)
train['Has2ndFloor'] = (train['2ndFlrSF'] > 0).astype(int)
train['HasBsmt'] = (train['TotalBsmtSF'] > 0).astype(int)
```

```python
new_features = ['TotalSF', 'TotalBathrooms', 'HouseAge', 'RemodAge', 'GarageAge',
                 'TotalPorchArea', 'TotalQualityScore', 'QualityLivingArea',
                 'HasPool', 'HasGarage', 'HasFireplace', 'Has2ndFloor', 'HasBsmt']

print(train[new_features].describe())
print("\nAny negative ages?")
print((train[['HouseAge', 'RemodAge']] < 0).sum())
```

```python
correlations = train[new_features + ['SalePrice']].corr()['SalePrice'].sort_values(ascending=False)
print(correlations)
```

```python
train.to_csv("../data/processed/train_features.csv", index=False)
```
