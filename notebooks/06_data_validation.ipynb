```python
import pandas as pd
import numpy as np

train = pd.read_csv("../data/processed/train_cleaned.csv")

required_columns = [
    "SalePrice",
    "GrLivArea",
    "OverallQual",
    "YearBuilt",
    "Neighborhood"
]

missing_columns = [
    col for col in required_columns
    if col not in train.columns
]

assert not missing_columns, f"Missing columns: {missing_columns}"
assert train["SalePrice"].notna().all(), "SalePrice contains missing values"
assert (train["SalePrice"] > 0).all(), "SalePrice contains invalid values"
assert train["Id"].is_unique, "Duplicate Id values found"

print("Rows:", train.shape[0])
print("Columns:", train.shape[1])
print("Duplicate rows:", train.duplicated().sum())
print("Missing values:", train.isna().sum().sum())
print("Data validation passed.")
```

```python
X_train = pd.read_csv("../data/processed/X_train.csv")
X_test = pd.read_csv("../data/processed/X_test.csv")

assert list(X_train.columns) == list(X_test.columns), \
    "Train and test columns do not match"

assert set(X_train.columns) == set(X_test.columns), \
    "Train and test feature sets are different"

print("Train/test schema validation passed.")
```
